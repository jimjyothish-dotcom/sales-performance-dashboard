# Supermart EDA

Exploratory data analysis on the Sample Superstore dataset.

This notebook contains data cleaning steps, feature engineering, and basic EDA
on the retail dataset.

```python
import pandas as pd
import numpy as np
```

```python
# Load the raw dataset
df = pd.read_csv('data/raw/Sample - Superstore.csv', encoding='latin1')

# Display the first few rows
df.head()
```

```python
# Clean column names: strip whitespace and lowercase
df.columns = df.columns.str.strip().str.lower()

# Parse dates
df['order date'] = pd.to_datetime(df['order date'], format='%m/%d/%Y')
df['ship date'] = pd.to_datetime(df['ship date'], format='%m/%d/%Y')
```

```python
# Derive time-based features
df['year'] = df['order date'].dt.year
df['month'] = df['order date'].dt.month
df['yearmonth'] = df['order date'].dt.to_period('M').astype(str)
df['ship_days'] = (df['ship date'] - df['order date']).dt.days
df['profit_margin'] = df['profit'] / df['sales']
```

```python
# Create loss and discount flags
df['IsLoss'] = (df['profit'] < 0).astype(int)
df['HighDiscount'] = (df['discount'] >= 0.5).astype(int)

# Print summary counts
print(f'Loss rows:          {df["IsLoss"].sum():,}')
print(f'High discount rows: {df["HighDiscount"].sum():,}')
```

```python
# Standardize string columns
str_cols = df.select_dtypes(include='object').columns
for col in str_cols:
    df[col] = df[col].str.strip()

for col in ['region', 'category', 'sub-category', 'segment', 'ship mode']:
    df[col] = df[col].str.title()
```

```python
# Export cleaned dataset
df.to_csv('data/processed/superstore_cleaned.csv', index=False)
print('Saved superstore_cleaned.csv')
```
